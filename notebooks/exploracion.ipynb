```python
import pandas as pd 

RUTA = "../data/raw/소상공인시장진흥공단_상가(상권)정보_경기_202606.csv"

try:
    df = pd.read_csv(RUTA, encoding='utf-8-sig', nrows=50000, low_memory=False)  # Cargar solo las primeras 50000 filas para exploración
    print("Archivo CSV cargado exitosamente.")
except UnicodeDecodeError:
    try:
        df = pd.read_csv(RUTA, encoding='cp949', nrows=50000, low_memory=False)  # Cargar solo las primeras 50000 filas para exploración
        print("Archivo CSV cargado exitosamente.")
    except FileNotFoundError:
        print(f"Error: No se encontró el archivo en la ruta especificada: {RUTA}")
    except pd.errors.ParserError:
        print("Error: Hubo un problema al analizar el archivo CSV.")
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")
```

```python
display(df.head())
```

```python
print("Información del DataFrame:")
df.info()
print(f"Forma del DataFrame: {df.shape}")
print("-" * 40)
```

```python
print("Identifadores únicos de la columna '상가업소번호':")
if '상가업소번호' in df.columns:
    unique_ids = df['상가업소번호'].nunique()
    print(f"Número de identificadores únicos: {unique_ids}")
```

```python
print("Valores nulos por columna:")
print(df.isnull().sum())
print("-" * 40)
print("Porcentaje de valores nulos por columna:")
null_percentage = (df.isnull().sum() / len(df)) * 100
print(null_percentage)
print("-" * 40)
```

```python
duplicados = df.duplicated().sum()
print(f"Número de filas duplicadas: {duplicados}")
print("-" * 40)
print("Duplicados exactos por columna:")
for column in df.columns:
    duplicated_count = df[column].duplicated().sum()
    print(f"{column}: {duplicated_count}")
```

```python
print("Estadísticas descriptivas del DataFrame:")
display(df.describe())
print("-" * 40)
print("Estadisticas espaciales y geográficas del DataFrame:")
if '위도' in df.columns and '경도' in df.columns:
    print(f"Rango de latitud: {df['위도'].min()} a {df['위도'].max()}")
    print(f"Rango de longitud: {df['경도'].min()} a {df['경도'].max()}")
```

```python
values_counts = df.nunique()
print("Número de valores únicos por columna:")
for column, count in values_counts.items():
    print(f"{column}: {count}")
```

```python
dtype_counts = df.dtypes.value_counts()
print("Tipos de datos y su conteo:")
for dtype, count in dtype_counts.items():
    print(f"{dtype}: {count}")
```

```python
top = (
    df.groupby('상권업종대분류명')
    .size()
    .sort_values(ascending=False)
    .head(10)
)
print("Top 10 categorías de '상권업종대분류명' con más registros:")
print(top)
```
